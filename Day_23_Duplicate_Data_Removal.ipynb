# Day 23: Duplicate Data Removal

## Objective
Identify and remove duplicate records from a dataset using Pandas.

## Tools Used
- Python
- Pandas
- Jupyter Notebook

## Description
This task demonstrates how to identify duplicate rows, count duplicate records, remove duplicates, and compare the dataset before and after cleaning.

```python
import pandas as pd
```

```python
data = {
    "Name": [
        "Arun",
        "Priya",
        "Karthik",
        "Divya",
        "Arun",
        "Priya",
        "Rahul",
        "Karthik"
    ],
    "Age": [
        21, 22, 23, 21, 21, 22, 24, 23
    ],
    "Department": [
        "ECE",
        "CSE",
        "EEE",
        "ECE",
        "ECE",
        "CSE",
        "IT",
        "EEE"
    ]
}

df = pd.DataFrame(data)

df
```

```python
print("Original number of rows:", len(df))
print("Original number of columns:", len(df.columns))
```

```python
duplicate_rows = df[df.duplicated()]

print("Number of duplicate records:", df.duplicated().sum())

duplicate_rows
```

```python
df_cleaned = df.drop_duplicates()

df_cleaned
```

```python
print("Before cleaning:", len(df), "rows")
print("After cleaning:", len(df_cleaned), "rows")
print("Rows removed:", len(df) - len(df_cleaned))
print("Remaining duplicate rows:", df_cleaned.duplicated().sum())
```

```python
df_cleaned.to_csv("duplicate_data_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```

```python
cleaned_data = pd.read_csv("duplicate_data_cleaned.csv")

print("Saved dataset shape:", cleaned_data.shape)
print("Duplicate rows in saved dataset:", cleaned_data.duplicated().sum())

cleaned_data
```

## Conclusion

Duplicate data removal was successfully performed using Pandas. The original dataset contained 8 rows, including 3 duplicate records. After applying `drop_duplicates()`, the cleaned dataset contained 5 unique rows. The cleaned dataset was exported to a CSV file for further analysis.
